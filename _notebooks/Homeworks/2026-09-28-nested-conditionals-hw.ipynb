## Popcorn Hack 1: After-School Homework Plan

Prediction for the starter values: **Finish homework**.
The weekday condition is True, so Python enters the outer branch.
The homework condition is False, so the inner else runs.

The code tests all three outcomes: Free time, Finish homework,
and Weekend plan.

````python
# CODE_RUNNER: Popcorn Hack 1 - After-school homework plan

test_cases = [
    (True, True),
    (True, False),
    (False, False)
]

for is_weekday, homework_finished in test_cases:
    print(f"\nWeekday: {is_weekday}, homework finished: {homework_finished}")

    # The homework check only runs inside the weekday branch.
    if is_weekday:
        if homework_finished:
            print("Free time")
        else:
            print("Finish homework")
    else:
        print("Weekend plan")
````

## Popcorn Hack 2: Repair the Student Account Check

The original permission check was aligned with the login check,
so it ran independently. Indenting it inside the logged-in branch
makes it a nested conditional.

The outer else handles students who are not logged in.
The tests cover Page opened, Permission needed, and Log in first.

````python
# CODE_RUNNER: Popcorn Hack 2 - Repair the student account check

test_cases = [
    (True, True),
    (True, False),
    (False, False)
]

for logged_in, has_permission in test_cases:
    print(f"\nLogged in: {logged_in}, permission: {has_permission}")

    # Check permissions only after login succeeds.
    if logged_in:
        print("Account found")
        if has_permission:
            print("Page opened")
        else:
            print("Permission needed")
    else:
        print("Log in first")
````

## Popcorn Hack 3: Lunch and Sports Practice

The outer condition checks whether lunch is packed.
The selected lunch branch then checks whether practice is scheduled.
All four Boolean combinations are tested.

````python
# CODE_RUNNER: Popcorn Hack 3 - Lunch and sports practice

test_cases = [
    (True, True),
    (True, False),
    (False, True),
    (False, False)
]

for lunch_packed, practice_after_school in test_cases:
    print(f"\nLunch packed: {lunch_packed}, practice: {practice_after_school}")

    if lunch_packed:
        if practice_after_school:
            print("Bring packed lunch and an extra practice snack")
        else:
            print("Bring packed lunch")
    else:
        if practice_after_school:
            print("Buy lunch and get an extra snack for practice")
        else:
            print("Buy lunch")
````

## Homework Hack: High School Daily Routine Dashboard

The dashboard stores login, homework, lunch, and practice status.
Login is checked before any personalized schedule recommendations.
Nested decisions then provide homework, meal, and practice advice.

````python
# CODE_RUNNER: Homework Hack - Daily routine dashboard

student = {
    "logged_in": True,
    "homework_finished": False,
    "lunch_packed": True,
    "practice_today": True
}


def daily_routine(student):
    # Schedule checks belong inside the login branch.
    if student["logged_in"]:
        print("Account access granted")

        if student["homework_finished"]:
            print("Homework complete — after-school time is available")
        else:
            print("Finish homework before free time")

        # Practice changes the meal recommendation within each lunch branch.
        if student["lunch_packed"]:
            if student["practice_today"]:
                print("Bring packed lunch and an extra snack for practice")
            else:
                print("Bring packed lunch")
        else:
            if student["practice_today"]:
                print("Buy lunch and get an extra snack for practice")
            else:
                print("Buy lunch")

        if student["practice_today"]:
            if student["homework_finished"]:
                print("Go to practice after school")
            else:
                print("Plan homework time before practice")
        else:
            print("No practice today — use the extra time for schoolwork or rest")
    else:
        print("Log in first before viewing the daily routine")


daily_routine(student)
````

## Testing and Explanation

The tests below run the same dashboard function with all 16
combinations of the four Boolean values.

Schedule checks are inside the login branch so personalized
recommendations appear only after login succeeds. The practice
check inside each lunch branch determines whether an extra snack
is needed. The homework check inside the practice branch determines
whether the student is ready for practice or needs to plan homework time.

````python
# CODE_RUNNER: Homework tests - All dashboard input combinations

from itertools import product

for case_number, values in enumerate(
    product([False, True], repeat=4),
    start=1
):
    test_student = {
        "logged_in": values[0],
        "homework_finished": values[1],
        "lunch_packed": values[2],
        "practice_today": values[3]
    }

    print(f"\n--- Test {case_number} ---")
    print(test_student)
    daily_routine(test_student)
````
