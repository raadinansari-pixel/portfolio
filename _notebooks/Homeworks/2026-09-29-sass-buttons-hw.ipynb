## Popcorn Hack 1: Add the Base Button Class

The `ocs__btn` class gives the resource link the shared OCS button styling.
An anchor is appropriate because this action navigates to another page.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 1 - Base Button Class -->
<!-- The base class applies the shared OCS button styling. -->
<a class="ocs__btn"
   href="https://www.mygoodbrain.org/resources"
   target="_blank"
   rel="noopener">
  Mental Health Resource Directory
  </a>
```

## Popcorn Hack 2: Modifiers and Button Groups

The main action uses `accent`, `pill`, and `fill`.
Both actions are grouped inside an `ocs__links` container.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 2 - Modifiers and Button Group -->
<!-- The links container groups related actions with shared spacing. -->
<!-- Pill and fill emphasize the main check-in action. -->
<div class="ocs__links">
  <button type="button" class="ocs__btn accent pill fill">
    Complete Daily Check-In
  </button>
  <a class="ocs__btn outline"
     href="https://www.mygoodbrain.org/resources"
     target="_blank"
     rel="noopener">
    Save Draft Journal
  </a>
</div>
```

## Popcorn Hack 3: Status Tones and Wide Layout

The check-in action uses `alert-green outline`.
The crisis-support action uses `alert-red fill`.
The `ocs__links--wide` container supplies the wide action layout.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 3 - Status Tones and Wide Layout -->
<!-- Status classes communicate priority using the shared theme. -->
<!-- Fill emphasizes the crisis-support action. -->
<div class="ocs__links--wide">
  <button type="button" class="ocs__btn alert-green outline">
    Complete Check-In
  </button>
  <a class="ocs__btn alert-red fill"
     href="https://www.mygoodbrain.org/resources"
     target="_blank"
     rel="noopener">
    Get Crisis Support
  </a>
</div>
```

## Homework Hack 1: Adaptive Resource Toolbar

The toolbar uses OCS classes instead of inline styling.
Urgent Support uses red, Self-Care uses green, and Community
uses an accent outline. All three links use the pill modifier.

```python
%%html
<!-- UI_RUNNER: Homework Hack 1 - Adaptive Resource Toolbar -->
<!-- All three links use OCS classes without inline styles. -->
<!-- Color and fill modifiers establish the resource priorities. -->
<div class="ocs__links">
  <a class="ocs__btn alert-red fill pill"
     href="https://www.mygoodbrain.org/resources"
     target="_blank"
     rel="noopener">
    Urgent Support
  </a>

  <a class="ocs__btn alert-green fill pill"
     href="https://www.mygoodbrain.org/resources"
     target="_blank"
     rel="noopener">
    Self-Care
  </a>

  <a class="ocs__btn accent outline pill"
     href="https://www.mygoodbrain.org/resources"
     target="_blank"
     rel="noopener">
    Community
  </a>
</div>
```

## Homework Hack 2: Primary vs. Secondary Action Priority

Log Mood Entry uses `accent fill` as the primary action.
View Weekly Trends uses `outline` as a secondary action.
The Crisis Hotline label uses `alert-red outline`.

The links retain the starter resource-directory URL for this styling exercise.

```python
%%html
<!-- UI_RUNNER: Homework Hack 2 - Primary vs Secondary Action Priority -->
<!-- The filled accent button establishes the primary action. -->
<!-- Outline styling gives the other actions less visual emphasis. -->
<div class="ocs__links">
  <button type="button" class="ocs__btn accent fill">
    Log Mood Entry
  </button>

  <a class="ocs__btn outline"
     href="https://www.mygoodbrain.org/resources"
     target="_blank"
     rel="noopener">
    View Weekly Trends
  </a>

  <a class="ocs__btn alert-red outline"
     href="https://www.mygoodbrain.org/resources"
     target="_blank"
     rel="noopener">
    Crisis Hotline
  </a>
</div>
```
