## Popcorn Hack 1: Refactor the Beakers

One standard OCS grid contains three beakers. Each beaker uses
`ocs__grid-cell`, with no custom classes or inline styles.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 1 - Three OCS Beakers -->
<!-- The standard variant organizes the beakers into three columns. -->
<!-- Every direct child uses the required grid-cell class. -->
<div class="ocs__grid ocs__grid--standard">
  <div class="ocs__grid-cell">Beaker A: Water</div>
  <div class="ocs__grid-cell">Beaker B: Salt Water</div>
  <div class="ocs__grid-cell">Beaker C: Vinegar</div>
</div>
```

## Popcorn Hack 2: Size the Control Beaker

`1fr 1fr 1fr 2fr` creates four columns. The Control Beaker receives
twice the space of each other beaker. This hack explicitly permits
the inline grid-template-columns style.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 2 - Control Beaker Twice as Wide -->
<!-- Four fraction values place all four beakers in one row. -->
<!-- The final 2fr makes the control twice as wide as each 1fr column. -->
<div class="ocs__grid ocs__grid--standard" style="grid-template-columns: 1fr 1fr 1fr 2fr;">
  <div class="ocs__grid-cell">Beaker 1</div>
  <div class="ocs__grid-cell">Beaker 2</div>
  <div class="ocs__grid-cell">Beaker 3</div>
  div class="ocs__grid-cell">Control Beaker</div>
</div>
```

## Homework Hack: Temperature Trial and Calculator

The temperature grid uses the standard variant. Its header spans
the row, and Beaker 3 is accented as the outlier.

The keypad uses the calculator variant, with an accented Avg key.

Run the cell and verify that all eight checklist lines display PASS.
Check the grid styling on the published portfolio page.

```python
%%html
<!-- UI_RUNNER: Homework Hack - Temperature and Calculator Grids -->
<!-- The standard grid highlights the outlier with an accent cell. -->
<!-- The calculator variant groups keys and highlights the Avg key. -->

<div id="grids-homework-output"></div>

<script>
(function () {
  const myHTML = [
    '<div class="ocs__grid ocs__grid--standard">',
    '  <div class="ocs__grid-cell ocs__grid-cell--header">Temperature Trial (°C)</div>',
    '  <div class="ocs__grid-cell">Beaker 1: 22.4</div>',
    '  <div class="ocs__grid-cell">Beaker 2: 22.9</div>',
    '  <div class="ocs__grid-cell ocs__grid-cell--accent">Beaker 3: 31.6</div>',
    '</div>',
    '<br>',
    '<div class="ocs__grid ocs__grid--calculator">',
    '  <div class="ocs__grid-cell">7</div>',
    '  <div class="ocs__grid-cell">8</div>',
    '  <div class="ocs__grid-cell">9</div>',
    '  <div class="ocs__grid-cell ocs__grid-cell--accent">Avg</div>',
    '</div>'
  ].join('');

  const usedClasses = [];

  for (const attr of (myHTML.match(/class="[^"]*"/g) || [])) {
    for (const name of attr.slice(7, -1).split(' ')) {
      if (name) usedClasses.push(name);
    }
  }

  const madeUp = usedClasses.filter(
    name => !name.startsWith('ocs__') && !name.startsWith('cols-')
  );

  const countOf = target =>
    usedClasses.filter(name => name === target).length;

  function row(label, passes) {
    return '<li>' + (passes ? 'PASS: ' : 'FIX: ') + label + '</li>';
  }

  const report = '<ul>' +
    row('No inline style attributes', !myHTML.includes('style=')) +
    row('No made up classes', madeUp.length === 0) +
    row('Has an ocs__grid root', countOf('ocs__grid') > 0) +
    row('Uses the standard variant for Grid 1',
        countOf('ocs__grid--standard') > 0) +
    row('Uses the calculator variant for Grid 2',
        countOf('ocs__grid--calculator') > 0) +
    row('Every box uses ocs__grid-cell',
        countOf('ocs__grid-cell') >= 8) +
    row('Title is a header cell',
        countOf('ocs__grid-cell--header') > 0) +
    row('Has accent cells for the outlier and Avg key',
        countOf('ocs__grid-cell--accent') >= 2) +
    '</ul>';

  document.getElementById('grids-homework-output').innerHTML =
    report + myHTML;
})();
</script>
```
