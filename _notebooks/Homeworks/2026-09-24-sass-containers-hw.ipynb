## Popcorn Hack: Parade Invite Rescue

The parade invite uses one `ocs__container` holding one `ocs__card`.
The title and description use OCS classes, and the schedule uses
a standard three-column grid with an accented start time.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack - Parade Invite Rescue -->
<!-- The container groups the invite within the shared page layout. -->
<!-- Each schedule item uses a grid cell instead of a custom flex layout. -->

<div class="ocs__container">
  <div class="ocs__card">
    <h2 class="ocs__section-title">Poway Rotary Parade Invite</h2>
    <p class="ocs__description">Join the Poway Scripps Rotary Club for a morning of community, service, and parade fun.</p>

    <div class="ocs__grid ocs__grid--standard cols-3">
      <div class="ocs__grid-cell">Location: Poway Road</div>
      <div class="ocs__grid-cell ocs__grid-cell--accent">Start Time: 9:00 AM</div>
      <div class="ocs__grid-cell">What You Get: Community fun and Rotary spirit</div>
    </div>
  </div>
</div>
```

## Homework Hack: Parade Hub Page

The parade hub uses exactly one container and one card. It includes
a title, description, standard three-column grid with header,
accent, and muted cells, a table with a header and two body rows,
and a final callout.

The page uses OCS classes without custom CSS or inline styles.

```python
%%html
<!-- UI_RUNNER: Homework Hack - Poway Rotary Parade Hub -->
<!-- One container holds the entire hub for consistent page spacing. -->
<!-- The grid highlights key details, while the table organizes event status. -->

<div class="ocs__container">
  <div class="ocs__card">
    <h2 class="ocs__section-title">Poway Scripps Rotary Parade Hub</h2>
    <p class="ocs__description">Find parade details, volunteer information, and event updates in one Rotary hub.</p>

    <div class="ocs__grid ocs__grid--standard cols-3">
      <div class="ocs__grid-cell ocs__grid-cell--header">Parade Day Information</div>
      <div class="ocs__grid-cell ocs__grid-cell--accent">Parade Start: 9:00 AM on Poway Road</div>
      <div class="ocs__grid-cell ocs__grid-cell--muted">Volunteer Check-In: 8:00 AM</div>
    </div>

    <div class="ocs__table-wrap">
      <table class="ocs__table">
        <thead>
          <tr>
            <th>Event</th>
            <th>Status</th>
          </tr>
        </thead>
        <tbody>
          <tr>
            <td>Parade Floats</td>
            <td>Confirmed</td>
          </tr>
          <tr>
            <td>Food Drive</td>
            <td>Collecting donations</td>
          </tr>
        </tbody>
      </table>
    </div>

    <div class="ocs__callout">Join the Poway Scripps Rotary Club in celebrating community and supporting local service projects.</div>
  </div>
</div>
```
