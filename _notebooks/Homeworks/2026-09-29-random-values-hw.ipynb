## Popcorn Hack 1: Random SFI Record Number

Prediction: the first selection can be any integer from 1 through 4.
Its exact value cannot be predicted.

randint includes both endpoints. Repeated values are allowed because
each selection is independent. Changing record_count to 5 makes
record number 5 reachable too.

```python
# CODE_RUNNER: Popcorn Hack 1 - Random SFI Record Number

import random

for record_count in [4, 5]:
    print(f"\nPossible record IDs: 1 through {record_count}")

    for test_number in range(5):
        record_id = random.randint(1, record_count)
        print("Random record ID:", record_id)
```

## Popcorn Hack 2: Random Structured SFI Part

Prediction: the output will contain the product name, category,
and spec number from one complete record.

random.choice selects a whole dictionary, keeping its fields together.
Adding another record expands the possible selections without changing
the selection algorithm.

```python
# CODE_RUNNER: Popcorn Hack 2 - Random SFI Part Record

import random

parts = [
    {
        "product_name": "Replacement Flywheels and Clutch Assemblies",
        "category": "Auto Racing",
        "spec_number": "1.1"
    },
    {
        "product_name": "Multiple Disc Clutch Assemblies",
        "category": "Drag Racing",
        "spec_number": "1.2"
    },
    {
        "product_name": "Racing Flywheel Record",
        "category": "Auto Racing",
        "spec_number": "2.1"
    }
]

def display_random_part(records):
    # Selecting the dictionary preserves the relationship between its fields.
    selected = random.choice(records)
    print("Product:", selected["product_name"])
    print("Category:", selected["category"])
    print("Spec number:", selected["spec_number"])

print("Original list:")
display_random_part(parts)

parts.append({
    "product_name": "Additional QA Test Part",
    "category": "Drag Racing",
    "spec_number": "3.1"
})

print("\nExpanded list:")
display_random_part(parts)
```

## Popcorn Hack 3: Random SFI API Test

Prediction: searching for spec 1.1 will find a stored record.
Creating spec 1.1 will be rejected because it already exists.

The random test selects both a record and an action.
Controlled tests then demonstrate each action without relying on
random selection to cover every branch.

```python
# CODE_RUNNER: Popcorn Hack 3 - Random SFI API Test

import random

parts = [
    {"product_name": "Replacement Flywheels", "spec_number": "1.1"},
    {"product_name": "Multiple Disc Clutch Assemblies", "spec_number": "1.2"},
    {"product_name": "Racing Flywheel Record", "spec_number": "2.1"}
]

actions = ["GET/search", "POST/create", "PUT/update"]
existing_spec_numbers = ["1.1", "2.1"]


def simulate_action(part, action):
    spec = part["spec_number"]

    if action == "GET/search":
        if spec in existing_spec_numbers:
            return "SEARCH: record found"
        return "SEARCH: no matching record"

    elif action == "POST/create":
        if spec in existing_spec_numbers:
            return "CREATE REJECTED: duplicate spec"
        return "CREATE ACCEPTED: new spec"

    elif action == "PUT/update":
        if spec in existing_spec_numbers:
            return "UPDATE ACCEPTED: existing record"
        return "UPDATE REJECTED: record does not exist"

    return "REJECTED: unknown action"


part = random.choice(parts)
action = random.choice(actions)

print("Random selection:", part["product_name"], "|", action)
print(simulate_action(part, action))

print("\nControlled action tests:")
for action in actions:
    for part in parts[:2]:
        print(part["spec_number"], action, simulate_action(part, action))
```

## Homework Hack: SFI Backend QA Simulator

The reusable simulator randomly selects complete part records and
backend actions. It runs five tests and reports each selected record,
action, and result.

Creation rejects duplicate specs and stores new ones. Search and update
check whether a spec exists. Delete removes an existing spec.

Each simulator run starts with its own copy of the initial stored specs.

```python
# CODE_RUNNER: 3.15 Homework - SFI Backend QA Simulator

import random

parts = [
    {
        "product_name": "Replacement Flywheels",
        "category": "Auto Racing",
        "spec_number": "1.1"
    },
    {
        "product_name": "Multiple Disc Clutch Assemblies",
        "category": "Drag Racing",
        "spec_number": "1.2"
    },
    {
        "product_name": "Racing Flywheel Record",
        "category": "Auto Racing",
        "spec_number": "2.1"
    }
]

actions = ["GET/search", "POST/create", "PUT/update", "DELETE/remove"]
existing_spec_numbers = ["1.1", "2.1"]
test_count = 5


def backend_result(part, action, stored_specs):
    """Simulate an action and update stored specs for create/delete."""
    spec = part["spec_number"]

    if action == "GET/search":
        if spec in stored_specs:
            return "SEARCH: record found"
        return "SEARCH: no matching record"

    elif action == "POST/create":
        if spec in stored_specs:
            return "CREATE REJECTED: duplicate spec number"
        stored_specs.add(spec)
        return "CREATE ACCEPTED: spec stored"

    elif action == "PUT/update":
        if spec in stored_specs:
            return "UPDATE ACCEPTED: existing record can be updated"
        return "UPDATE REJECTED: record does not exist"

    elif action == "DELETE/remove":
        if spec in stored_specs:
            stored_specs.remove(spec)
            return "DELETE ACCEPTED: spec removed"
        return "DELETE REJECTED: record does not exist"

    return "REJECTED: unknown action"


def run_qa(records, available_actions, initial_specs, runs):
    """Run randomized tests and return the final set of stored specs."""
    # Copy the initial state so separate simulator runs do not affect each other.
    stored_specs = set(initial_specs)

    for test_number in range(1, runs + 1):
        # Choose the entire record so its fields remain connected.
        part = random.choice(records)
        action = random.choice(available_actions)

        print(f"\nTest {test_number}")
        print("Product:", part["product_name"])
        print("Category:", part["category"])
        print("Spec:", part["spec_number"])
        print("Action:", action)
        print("Result:", backend_result(part, action, stored_specs))

    print("\nFinal stored specs:", sorted(stored_specs))
    return stored_specs


run_qa(parts, actions, existing_spec_numbers, test_count)

# Controlled tests guarantee that important behaviors are demonstrated.
print("\n--- Controlled tests ---")
stored_specs = set(existing_spec_numbers)
new_part = parts[1]

print("Search existing:",
      backend_result(parts[0], "GET/search", stored_specs))
print("Reject duplicate:",
      backend_result(parts[0], "POST/create", stored_specs))
print("Update missing:",
      backend_result(new_part, "PUT/update", stored_specs))
print("Create new:",
      backend_result(new_part, "POST/create", stored_specs))
print("Reject repeated creation:",
      backend_result(new_part, "POST/create", stored_specs))
print("Update existing:",
      backend_result(new_part, "PUT/update", stored_specs))
print("Delete existing:",
      backend_result(new_part, "DELETE/remove", stored_specs))
print("Search after deletion:",
      backend_result(new_part, "GET/search", stored_specs))
print("Delete missing:",
      backend_result(new_part, "DELETE/remove", stored_specs))
```

## Randomness and Pseudocode Comparison

With three records and four actions, there are 3 × 4 = 12 possible
record/action pairs. Pairs can repeat, and five random tests do not
guarantee that every pair will appear. Controlled tests demonstrate
specific behaviors that random runs might miss.

College Board RANDOM(a, b) and Python random.randint(a, b) both include
their endpoints. College Board list positions start at 1, while Python
list indexes start at 0. random.choice selects a record directly,
avoiding manual index conversion.
