## Popcorn Hack 1: Read the API, Then Use It

Prediction: the jersey number will be between 1 and 55, inclusive.
The selected game score will be one of the values in the list.
The square root of 225 is 15.0, and the average of the scores
is 24.4, which rounds up to 25.

````python
# CODE_RUNNER: 3.14 Popcorn 1 - Math and random APIs

import math
import random

# Documentation explains the parameters and return values.
print("random.randint:", random.randint.__doc__)
print("math.ceil:", math.ceil.__doc__)
print("math.sqrt:", math.sqrt.__doc__)
print("Available math names:", dir(math))
print("Available random names:", dir(random))

points_per_game = [24, 18, 31, 27, 22]

print("\nRandom game to review:",
      random.choice(points_per_game), "points")

# randint includes both endpoints.
jersey = random.randint(1, 55)
print("Random jersey number:", jersey)

side = math.sqrt(225)
print("Square court side:", side, "feet")

average = sum(points_per_game) / len(points_per_game)
print("Points per game:", average)
print("Average rounded up:", math.ceil(average))
````

# Popcorn Hack 2: Write a Flask Route

The roster includes Raadin and two example teammates.
A new /coach route returns coaching information.

All routes are defined before creating the test client.
The test client requests each route without starting a web server.

````python
# CODE_RUNNER: 3.14 Popcorn 2 - Customized Flask routes

from flask import Flask, jsonify

app = Flask(__name__)


@app.route("/score")
def score():
    return jsonify({"home": 6, "away": 4, "set": 3})


@app.route("/roster")
def roster():
    return jsonify({"players": ["Raadin", "Alex", "Jordan"]})


# A decorator connects this procedure to its web address.
@app.route("/coach")
def coach():
    return jsonify({
        "coach": "Coach Patel",
        "focus": "serve consistency"
    })


# Test the real Flask routes without running a separate server.
with app.test_client() as client:
    for route in ["/score", "/roster", "/coach"]:
        response = client.get(route)
        print(route, "status:", response.status_code)
        print("Response:", response.get_json())
````

## Homework Hack: Tennis Season Stats Report

The report uses json, statistics, datetime, and math.

The summarize_season procedure returns total points, average
points, the highest single-game score, and points grouped by opponent.
Its docstring documents the parameters and return value.

The report also includes days until the next game and total match
time rounded up to whole hours.

````python
# CODE_RUNNER: 3.14 Homework - Tennis season stats report

import json
import math
import statistics
from datetime import date

# Example data: replace with your own games before submitting.
games = [
    {"date": "2026-09-05", "opponent": "Westview", "points": 12, "minutes": 64},
    {"date": "2026-09-12", "opponent": "Del Norte", "points": 15, "minutes": 71},
    {"date": "2026-09-19", "opponent": "Westview", "points": 10, "minutes": 58},
    {"date": "2026-09-26", "opponent": "Poway", "points": 18, "minutes": 83}
]


def summarize_season(season):
    """Summarize a list of game dictionaries.

    Parameters:
        season: List of dictionaries with date, opponent,
                points, and minutes fields.

    Returns:
        Dictionary containing total_points, average_points,
        best_game (highest single-game score), and
        points_by_opponent.
        An empty season returns zero scores and an empty
        opponent dictionary.
    """
    points = [game["points"] for game in season]

    # Combine scores when an opponent appears more than once.
    points_by_opponent = {}
    for game in season:
        opponent = game["opponent"]
        points_by_opponent[opponent] = (
            points_by_opponent.get(opponent, 0) + game["points"]
        )

    return {
        "total_points": sum(points),
        "average_points": (
            round(statistics.mean(points), 1) if points else 0
        ),
        "best_game": max(points) if points else 0,
        "points_by_opponent": points_by_opponent
    }


report = summarize_season(games)

# Replace this example with your chosen next-game date.
next_game = date(2026, 10, 15)
report["next_game_date"] = next_game.isoformat()
report["days_until_next_game"] = max(
    0, (next_game - date.today()).days
)

# ceil rounds the total match time upward to whole hours.
total_minutes = sum(game["minutes"] for game in games)
report["match_hours_rounded_up"] = math.ceil(total_minutes / 60)

# JSON makes the report readable and suitable for a web response.
print(json.dumps(report, indent=2))

requirements_txt = """pandas
Flask
# json, datetime, statistics, and math come with Python.
# Standard-library modules do not need pip installation.
"""

print("\nrequirements.txt:")
print(requirements_txt)
````
