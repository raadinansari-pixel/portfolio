## Popcorn Hack 1: Read the API, Then Use It

I changed the jersey range to 1–55 and the square court area to 225 square feet.

Prediction: the random jersey will be between 1 and 55, including both endpoints. The selected score will be one of the list values. The square root will be 15.0. The average will be 24.4, and math.ceil will round it up to 25.

The API documentation explains the parameters and return values. dir() lists the names available in each module.

````python
# CODE_RUNNER: 3.14 Popcorn Hack 1 - Math and Random APIs

import math
import random

print("random.randint documentation:", random.randint.__doc__)
print("math.sqrt documentation:", math.sqrt.__doc__)
print("math.ceil documentation:", math.ceil.__doc__)
print("Available math names:", dir(math))
print("Available random names:", dir(random))

points_per_game = [24, 18, 31, 27, 22]

# choice selects one value from the existing list.
print("\nRandom game to review:",
      random.choice(points_per_game), "points")

# randint includes both endpoints.
print("Random jersey number:", random.randint(1, 55))

print("Square court side:", math.sqrt(225), "feet")

average = sum(points_per_game) / len(points_per_game)
print("Average points:", average)
print("Average rounded up:", math.ceil(average))
````

## Popcorn Hack 2: Write a Flask Route

My chosen roster contains Raadin, Alex, and Jordan. I added a /coach route that returns the coach's name and practice focus.

The route decorator connects a URL to a procedure. jsonify converts the returned data into a JSON response.

Prediction: /score, /roster, and /coach will each return HTTP status 200 and their corresponding JSON data. All routes are defined before the test client is created.

````python
# CODE_RUNNER: 3.14 Popcorn Hack 2 - Customized Flask Routes

from flask import Flask, jsonify

app = Flask(__name__)


@app.route("/score")
def score():
    return jsonify({"home": 6, "away": 4, "set": 3})


@app.route("/roster")
def roster():
    return jsonify({"players": ["Raadin", "Alex", "Jordan"]})


# This additional route provides coaching information.
@app.route("/coach")
def coach():
    return jsonify({
        "coach": "Coach Patel",
        "focus": "serve consistency"
    })


# The test client requests real routes without starting a server.
with app.test_client() as client:
    for route in ["/score", "/roster", "/coach"]:
        response = client.get(route)
        print(route, "status:", response.status_code)
        print("Response:", response.get_json())
````

## Homework Hack: Tennis Season Stats Report

I chose tennis and created a fictional four-match dataset for this report. The points represent points won during each match. My chosen next-match date is October 20, 2026.

The program uses four standard-library modules:
- statistics calculates the average points.
- datetime calculates days until the next match.
- math rounds total match time up to whole hours.
- json formats the report as readable JSON.

The summarize_season procedure returns total points, average points, the highest single-match score, and points grouped by opponent. Its docstring explains the parameters and return value.

Prediction: total points will be 52, average points will be 13.0, and the best match score will be 17. Points by opponent will be Westview: 31, Poway: 11, and Rancho Bernardo: 10. Total match time is 250 minutes, which rounds up to 5 hours. Days until the next match depend on the date when the code runs.

````python
# CODE_RUNNER: 3.14 Homework - Tennis Season Stats Report

import json
import statistics
import math
from datetime import date

# My chosen fictional tennis match data.
games = [
    {
        "date": "2026-09-06",
        "opponent": "Westview",
        "points": 14,
        "minutes": 62
    },
    {
        "date": "2026-09-13",
        "opponent": "Poway",
        "points": 11,
        "minutes": 55
    },
    {
        "date": "2026-09-20",
        "opponent": "Westview",
        "points": 17,
        "minutes": 75
    },
    {
        "date": "2026-09-27",
        "opponent": "Rancho Bernardo",
        "points": 10,
        "minutes": 58
    }
]


def summarize_season(season):
    """Return summary statistics for a season.

    Parameters:
        season: A list of game dictionaries. Each dictionary has
                date and opponent strings, numeric points,
                and numeric minutes.

    Returns:
        A dictionary containing:
        total_points: Sum of points from all games.
        average_points: Mean points, rounded to one decimal place.
        best_game: Highest single-game point total.
        points_by_opponent: Total points grouped by opponent.

        An empty season returns zero totals and an empty
        points_by_opponent dictionary.
    """
    points = [game["points"] for game in season]

    # Repeated opponents share one accumulated total.
    points_by_opponent = {}
    for game in season:
        opponent = game["opponent"]
        points_by_opponent[opponent] = (
            points_by_opponent.get(opponent, 0) + game["points"]
        )

    return {
        "total_points": sum(points),
        "average_points": (
            round(statistics.mean(points), 1) if points else 0
        ),
        "best_game": max(points) if points else 0,
        "points_by_opponent": points_by_opponent
    }


report = summarize_season(games)

# My chosen next-match date.
next_game = date(2026, 10, 20)
today = date.today()

report["report_date"] = today.isoformat()
report["next_game_date"] = next_game.isoformat()
report["days_until_next_game"] = max(
    0, (next_game - today).days
)

# Round total playing time upward to whole hours.
total_minutes = sum(game["minutes"] for game in games)
report["total_match_minutes"] = total_minutes
report["match_hours_rounded_up"] = math.ceil(total_minutes / 60)

print(json.dumps(report, indent=2))
````
