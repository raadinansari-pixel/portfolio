## Homework Hack: Interactive Settings Panel

Three labeled OCS toggles control different features:

- Show Notifications reveals a notification message.
- Focus Mode hides a distracting message.
- Show Study Tip reveals a study tip.

The status counter updates whenever a toggle changes.

```python
%%html
<!-- UI_RUNNER: Homework Hack - Interactive Toggle Settings -->

<div id="toggles-homework" class="ocs__container">
  <div class="ocs__card">
    <h2 class="ocs__section-title">My Website Settings</h2>
    <p class="ocs__description">Use the switches to change this panel.</p>

    <!-- Each label wraps its checkbox and text, making both clickable. -->
    <div>
      <label class="ocs__toggle ocs__toggle--green">
        <input id="hw-notifications-toggle"
               class="ocs__toggle-input"
               type="checkbox">
        <span class="ocs__toggle-track" aria-hidden="true"></span>
        <span class="ocs__toggle-label">Show Notifications</span>
      </label>
    </div>

    <div>
      <label class="ocs__toggle ocs__toggle--green">
        <input id="hw-focus-toggle"
               class="ocs__toggle-input"
               type="checkbox">
        <span class="ocs__toggle-track" aria-hidden="true"></span>
        <span class="ocs__toggle-label">Focus Mode</span>
      </label>
    </div>

    <div>
      <label class="ocs__toggle ocs__toggle--green">
        <input id="hw-tips-toggle"
               class="ocs__toggle-input"
               type="checkbox">
        <span class="ocs__toggle-track" aria-hidden="true"></span>
        <span class="ocs__toggle-label">Show Study Tip</span>
      </label>
    </div>

    <div class="ocs__callout">
      Toggles currently on:
      <output id="hw-toggle-count" aria-live="polite">0 of 3</output>
    </div>

    <p id="hw-notification-message" hidden>
      🔔 Reminder: Check your homework deadlines.
    </p>

    <p id="hw-distraction">
      🎮 Distraction: Take a look at the latest game updates!
    </p>

    <p id="hw-focus-message" hidden>
      Focus Mode is on. The distracting message is hidden.
    </p>

    <p id="hw-study-tip" hidden>
      💡 Study Tip: Break a large task into smaller steps.
    </p>
  </div>
</div>

<script>
(function () {
  const panel = document.getElementById("toggles-homework");
  const toggles = panel.querySelectorAll(".ocs__toggle-input");

  const notifications = panel.querySelector("#hw-notifications-toggle");
  const focus = panel.querySelector("#hw-focus-toggle");
  const tips = panel.querySelector("#hw-tips-toggle");

  const notificationMessage = panel.querySelector("#hw-notification-message");
  const distraction = panel.querySelector("#hw-distraction");
  const focusMessage = panel.querySelector("#hw-focus-message");
  const studyTip = panel.querySelector("#hw-study-tip");
  const countOutput = panel.querySelector("#hw-toggle-count");

  function updatePanel() {
    // .checked is a Boolean that tells us whether each switch is on.
    // Setting hidden to the opposite value shows content only when enabled.
    notificationMessage.hidden = !notifications.checked;
    studyTip.hidden = !tips.checked;

    // Focus Mode removes an actual distraction from the panel.
    distraction.hidden = focus.checked;
    focusMessage.hidden = !focus.checked;

    // Count only this panel's switches so other page toggles cannot affect it.
    const count = Array.from(toggles).filter(toggle => toggle.checked).length;
    countOutput.textContent = count + " of 3";
  }

  toggles.forEach(toggle => {
    toggle.addEventListener("change", updatePanel);
  });

  // Synchronize the features and counter when the panel first loads.
  updatePanel();
})();
</script>
```
